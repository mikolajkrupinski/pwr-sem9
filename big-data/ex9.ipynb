```python
import random
from matplotlib import pyplot as plt
```

```python
def procedure(n: int, k: int) -> float:
    assert n >= 1 and k >= 1, "n and k must be positive integers"

    X = [[random.uniform(0, 1) for _ in range(n)] for _ in range(k)]

    d_max = 0.0
    d_min = float('inf')
    for i in range(k):
        for j in range(i + 1, k):
            dist = sum((X[i][m] - X[j][m]) ** 2 for m in range(n)) ** 0.5
            d_max = max(d_max, dist)
            d_min = min(d_min, dist)

    return d_max / d_min
```

```python
k = 50
n_values = [1, 10, 100, 1000, 10000]
results = {n: procedure(n, k) for n in n_values}
print(results)

plt.scatter(results.keys(), results.values())
plt.xscale('log')
plt.yscale('log')
plt.grid()
plt.xlabel('Dimension (n)')›
plt.ylabel('d_max / d_min')
plt.show()
```
