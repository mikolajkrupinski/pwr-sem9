```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('1.csv')
```

```python
plt.hist(df['sample-age'], bins=5, edgecolor='black', density=True)
```

```python
from scipy.stats import chisquare

counts = np.bincount(df['sample-age'], minlength=5)
N = len(df['sample-age'])

expected = [N / 5] * 5
chi2, p_value = chisquare(counts, expected)

print("Chi2:", chi2)
print("p-value:", p_value)
```
