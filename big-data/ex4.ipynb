```python
import math
from matplotlib import pyplot as plt
```

```python
def volume(radius: float, dimension: int) -> float:
    assert radius >= 0, "Radius must be non-negative"
    assert dimension > 0, "Dimension must be a positive integer"
    return ((math.pi ** (dimension / 2)) / math.gamma(dimension / 2 + 1)) * radius ** dimension
```

```python
dimension = range(1, 51)
radiuses = [0.5, 1, 2]
volumes = {r: [volume(r, d) for d in dimension] for r in radiuses}
```

```python
for r in radiuses:
    plt.scatter(dimension, volumes[r], label=f'Radius = {r}', s=10)
    plt.legend()
    plt.grid()
    plt.show()
```

```python
for r in radiuses:
    plt.scatter(dimension, volumes[r], label=f'Radius = {r}', s=10)
plt.grid()
plt.legend()
plt.show()
```

```python
for r in radiuses:
    plt.scatter(dimension, volumes[r], label=f'Radius = {r}', s=10)
plt.yscale('log')
plt.grid()
plt.legend()
plt.show()
```
